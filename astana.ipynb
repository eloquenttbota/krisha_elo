# 1. Введение и цели проекта

**Цель проекта** — применить знания и опыт, полученные на **AI Club
Practicum** (ведущая — Жанар Бакенова), на практике: создать рабочий
продукт — Telegram-бота для объективной оценки стоимости квартир в Астане.

Модель привносит рациональность в процесс сделки: **покупателю** она помогает 
увидеть рыночный ориентир цены и не переплачивать агентствам или продавцам, 
а **продавцу** — обосновать стоимость на основе параметров жилья, а не личных 
эмоций. При этом важно понимать: любая ML-модель имеет погрешность и не является 
абсолютной истиной. Бот создавался не для слепого следования цифрам, а как 
дополнительный инструмент, который стоит учитывать при финальном расчете 
стоимости перед покупкой или продажей.

**Задачи:**
1. Выбрать предметную область (недвижимость) и источник данных — сайт
   krisha.kz — и собрать датасет: парсинг объявлений о продаже квартир в Астане.
2. Понять данные (EDA) и проверить гипотезы о рынке недвижимости Астаны.
3. Очистить данные и восстановить пропуски.
4. Создать новые интерпретируемые признаки на основе логики рынка для повышения точности предсказания цены.
5. Обучить три разных алгоритма — **Decision Tree**, **Random Forest**,
   **XGBoost** — подобрать им параметры *без утечки данных* и сравнить по
   одной понятной метрике — **MAPE**, % ошибки.
6. Выбрать лучшую модель, сохранить её и подключить к Telegram-боту, который
   по введённым параметрам квартиры называет пользователю цену за секунды.

**Как устроен этот ноутбук.** Все вычисления вынесены в python-модули
в папке `src/`. Здесь, в ноутбуке, вызываются готовые функции — на экране
только результат: таблицы, графики, выводы.

<a id="cleaning-impute"></a>

### 5.5 Восстановление пропусков — двумя разными способами

Не всё стоит заполнять «по соседям на карте». Мы используем два разных
подхода в зависимости от природы признака:

* **Район** и **тип дома** — географически обусловлены, поэтому берём
  ближайшие по координатам объявления (k-NN).
* **Высота потолков** и **этажность дома** — определяются эпохой и
  материалом постройки, а не координатами: в одном районе спокойно
  соседствуют старая панельная пятиэтажка с потолками 2.5 м и новый
  монолитный дом с потолками 3.2 м. Поэтому здесь берём медиану среди
  объявлений **того же типа дома и того же десятилетия постройки**.


```python
df_clean = impute_district_by_location(df_clean)
df_clean = impute_house_type_knn(df_clean)
df_clean = impute_ceiling_height_by_type(df_clean)
df_clean = impute_floor_count_by_type(df_clean)
df_clean = finalize_cleaning(df_clean)
```

<a id="selection"></a>

# 7. Feature Selection

Три проверки перед обучением:
1. **Слабый сигнал** — признаки почти не связанные с ценой, убираем (кроме
   ключевых «понятных» признаков вроде площади — их оставляем всегда).
2. **Мультиколлинеарность** — если два признака дублируют друг друга
   (|Spearman r| > 0.90), оставляем только один.
3. **Утечка таргета** — проверяем, что цена не «просочилась» в признаки.


```python
from src.feature_selection import select_features

df_model = select_features(df_features, colors=COLORS)
```

### 8.2 Decision Tree — дерево решений

**🌲 Дерево решений** — это интуитивный алгоритм машинного обучения, который имитирует процесс принятия решений человеком. Вместо сложных математических формул модель строит логическую цепочку вопросов, на которые можно ответить «Да» или «Нет».

#### 🔍 Как это работает на практике (логика оценки квартиры):
1. **Корень (Главный критерий):** Модель выбирает параметр, который сильнее всего разделяет данные на «дорогие» и «дешевые» (например, *Площадь > 60 м²*).
2. **Ветви (Уточнения):** На каждом следующем шаге задаются сужающие вопросы (например, *Район = Левый берег?*, *Год постройки > 2015?*).
3. **Листья (Итоговое предсказание):** Конечные точки, где вопросы заканчиваются. В них находится средняя стоимость похожих квартир из обучающей выборки.

#### ⚖️ Особенности алгоритма в контексте задачи:
* **➕ Главный плюс — Интерпретируемость:** Это «прозрачный» алгоритм. Мы всегда можем проследить точную логику модели и понять, по каким именно критериям она сформировала предсказание стоимости.
* **⚠️ Главный риск — Переобучение (Overfitting):** Без жестких ограничений дерево может стать слишком глубоким, начав запоминать случайные шумы и мелкие детали объявлений. Это приведет к высокой ошибке на новых данных.

### 8.3 Random Forest — лес из деревьев

**🌲 Дерево решений** — это алгоритм, который прогнозирует цену квартиры с помощью последовательных вопросов, на которые можно ответить только **«Да»** или **«Нет»**. Он работает точно так же, как интерактивный тест в журнале или логика опытного риелтора.

#### 🛠 Как модель принимает решение:
1. **Задает главный вопрос:** Например, *«Площадь больше 60 кв.м?»* и разделяет данные на две ветки.
2. **Уточняет детали:** На каждой следующей ветке задает новые вопросы (про район, год постройки, количество комнат).
3. **Выдает ответ:** Когда вопросы заканчиваются, модель смотрит на среднюю цену похожих квартир в этой ветке и выдает её как предсказание.

#### 👍/👎 Что нужно знать об алгоритме:
* **Плюс (Прозрачность):** Мы всегда можем пошагово увидеть, почему модель оценила квартиру именно в эту сумму.
* **Минус (Капризность):** Если дерево сделать слишком большим, оно начнет цепляться к случайным мелочам и будет часто ошибаться на новых объявлениях.


```python
from src.tree_diagrams import draw_random_forest_html

draw_random_forest_html(COLORS)
```

```python
rf_model, rf_params = tune_random_forest(X_train, y_train, colors=COLORS)
```

### 8.4 XGBoost — градиентный бустинг

**🚀 XGBoost** — это один из самых мощных алгоритмов для работы с табличными данными. Если Случайный лес строит деревья одновременно и независимо, то XGBoost строит их **последовательно**, где каждое новое дерево проводит «работу над ошибками» предыдущих.

#### 🛠 Как модель принимает решение:
1. **Первая попытка:** Алгоритм строит первое, очень простое дерево. Оно предсказывает цену грубо и, конечно, ошибается.
2. **Анализ ошибок:** Модель смотрит, на каких именно квартирах первое дерево ошиблось сильнее всего (например, сильно занизило цену элитного жилья).
3. **Эстафета:** Второе дерево обучается специально для того, чтобы исправить именно эти ошибки первого. Третье исправляет ошибки второго, и так далее (до сотен шагов). 
4. **Финальный аккорд:** Итоговое предсказание — это сумма усилий всей цепочки деревьев, где каждое следующее докручивает точность до идеала.

#### 👍/👎 Что нужно знать об алгоритме:
* **Плюс (Максимальная точность):** Этот алгоритм выжимает из данных максимум. Именно с ним Data Scientist'ы чаще всего побеждают в соревнованиях на Kaggle, когда важен каждый процент точности.
* **Минус (Склонность к переобучению):** Из-за того, что модель фанатично исправляет каждую мелкую ошибку, её очень легко «перетренировать». Ей нужны четкие ограничения, иначе она начнет подгонять ответы под обучающую выборку.


```python
from src.tree_diagrams import draw_xgboost_html

draw_xgboost_html(COLORS)
```

```python
xgb_model, xgb_params = tune_xgboost(X_train, y_train, colors=COLORS)
```

### 8.5 Визуальное сравнение моделей

**📈 Кривые обучения (Learning Curves)** — это график-рентген, который показывает, как качество предсказания модели зависит от количества данных. Он помогает понять, «выучила» ли модель рыночную логику или просто зазубрила объявления, и нужно ли нам искать больше данных для улучшения точности.

На графике всегда рисуются две линии: ошибка на **обучающей** выборке (на чем модель тренировалась) и ошибка на **валидационной** выборке (новые квартиры, которые модель не видела).

#### 🛠 Как читать этот график через призму нашей метрики MAPE:
В нашей задаче мы оцениваем ошибку в процентах (**MAPE**). Чем ниже линия на графике — тем точнее предсказание и ближе к 0% ошибка.

1. **Если обе линии сошлись вверху (Высокая ошибка, например, MAPE > 25%):** 
   Модель слишком простая (недообучение). Ей дали 37 тысяч объявлений, но она не может уловить логику рынка. Сюда бесполезно добавлять новые данные — нужно усложнять саму модель (например, переходить от Дерева решений к XGBoost).
2. **Если линии далеко друг от друга (На обучении MAPE 2%, а на валидации — 20%):** 
   Модель переобучилась. Она просто «зазубрила» тренировочные квартиры до идеала, но пасует перед новыми. Ей нужны жесткие ограничения (меньше глубина дерева) либо больше разнообразных данных для тренировки.
3. **Идеальный финал (Линии близко друг к другу и внизу):** 
   Обе кривые стабилизировались на низком уровне ошибки. Это значит, что модель поймала баланс, отлично понимает логику рынка Астаны и готова к работе на реальных данных.

#### 👍/👎 Что нам дает этот инструмент:
* **Плюс (Экономия сил):** График четко показывает, имеет ли смысл дальше парсить *krisha.kz* ради новых строк. Если кривые уже выровнялись и идут параллельно, то приток новых данных ошибку больше не снизит — нужно менять подход к признакам.
* **Минус (Требует времени):** Чтобы построить этот график, модель приходится переобучать много раз, постепенно увеличивая порцию данных (10%, 20%, 30%... 100%), что требует вычислительных ресурсов.


```python
all_models = {"Decision Tree": dt_model, "Random Forest": rf_model, "XGBoost": xgb_model}
plot_learning_curves_grid(all_models, X_train, y_train, COLORS)
```
